```python
from dotenv import load_dotenv
load_dotenv()  # take environment variables from .env.
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings 
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
document = [
    "Artificial Intelligence mimics human intelligence to solve complex problems",
    "It powers virtual assistants like Siri, Alexa, and Google Assistant.",
    "Machine learning algorithms help AI learn from large datasets over time.",
    "AI can process and analyze massive amounts of data instantly.",
    "Self-driving cars use AI to navigate roads safely without drivers",
    "Healthcare systems use AI to detect diseases much faster than before",
    "Generative AI can create original text, art, and music instantly",
    "Automation through AI helps businesses perform repetitive tasks efficiently",
    "Smart recommendation systems use AI to suggest movies and products you like",
    "The technology continues to evolve and transform daily life globally",
    ]
```

```python
# vector = embeddings.embed_query("hello, world!")
vector = embeddings.embed_documents(document)
```

```python
print(len(vector))
print(len(vector[0]))
# vector[:5]
```

#### Store vector Embeddings into vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=document,
    embedding=embeddings,
    collection_name="ai_docs",
    persist_directory='../data/vector_chromaDB'
)
```

```python
query = "Artificial Intelligence"
result = vector_store.similarity_search(query=query, k=3)
result
```
