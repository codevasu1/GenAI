```python
from langchain_google_genai import ChatGoogleGenerativeAI
llmGoogle = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite",streaming=True)
```

```python
question = "can you explain me Gen-AI?"
```

### Streareaming response via llm

```python
res = llmGoogle.stream(question)
```

```python
for chunk in res:
    content = chunk.content

    if isinstance(content, str):
        print(content, end="", flush=True)
    elif isinstance(content, list):
        for item in content:
            if isinstance(item, dict) and item.get("type") == "text":
                print(item.get("text", ""), end="", flush=True)
```
