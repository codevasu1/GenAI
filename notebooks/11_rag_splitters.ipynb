 Document text loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
speech_loader = TextLoader('../data/speech.txt')
speech_texts = speech_loader.load()
```

```python
pdf_loader = PyPDFLoader('../data/sorting_algorithms_interview_guide.pdf')
pdf_texts = pdf_loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)
```

```python
chunks2 = splitter.split_documents(speech_texts)
```

```python
chunks3 = splitter.split_documents(pdf_texts)
```

```python
# chunks2
len( chunks2 )  # number of chunks created from the speech text
```

```python
# list of documents with metadata and text content, 
# each document is a chunk of the original PDF split according 
# to the specified chunk size and overlap.
len( chunks3 )  # number of chunks created from the PDF
# chunks3 
```
