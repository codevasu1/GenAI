### Text Document loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader('../data/speech.txt', encoding='utf-8')
texts = loader.load()

# shows number of documents loaded
len(texts)
oneDoc = texts[0]
print(oneDoc.page_content[:500])  # print first 500 characters of the document
```

### PDF Document loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('../data/sorting_algorithms_interview_guide.pdf')

docs = loader.load()
len(docs)
oneDoc = docs[0]
print(oneDoc.page_content[:500])  # print first 500 characters of the document
```
