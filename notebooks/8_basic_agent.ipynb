```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
from langchain_core.messages import AIMessage
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together.
    Args:
        a (int): The first number.
        b (int): The second number.
    Returns:
        int: The sum of the two numbers.
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers together.
    Args:
        a (int): The first number.
        b (int): The second number.
    Returns:
        int: The product of the two numbers.
    """
    return a * b
```

```python
add_numbers.invoke({"a": 12, "b": 3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llmGroq = ChatGroq(model="qwen/qwen3.8-27b")
```

```python
agent = create_agent(
    model=llmGroq,
    tools=[add_numbers, multiply_numbers],
    system_prompt="you are a math teacher and always use tool for calculation"
)
```

```python
# we pass dictonary and key must be 'messages'
res = agent.invoke({'messages':[{'role':'user', 'content':'what is 5 + 7 * 2?'}]}) 


# for straming the response
# res = agent.stream({'messages':[{'role':'user', 'content':'what is 3 + 7 * 2?'}]})
# contents = []

# for chunk in res:
#     for update in chunk.values():
#         for message in update.get("messages", []):
#             if isinstance(message, AIMessage) and message.content:
#                 contents.append(message.content)

# print("\n\n".join(contents))
```
