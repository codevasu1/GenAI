```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import AIMessage
```

```python
# llmGroq = ChatGroq(model="qwen/qwen3.8-27b")
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
GoogleSearch = GoogleSerperAPIWrapper()
```

```python
# search.run("Address of indian prime minister")
```

```python
agent = create_agent(
    model = llm,
    tools = [GoogleSearch.run],# notice: no colons after .run, we dont want to run it now
    system_prompt="You are a tool for searching the internet. you can search any question on google search.",
    checkpointer=MemorySaver()
)
```

```python
question = "total runs?"
```

```python
res = agent.invoke({'messages':[{'role': 'user', 'content': question}]},
                   {"configurable":{"thread_id":"45"}})
```

```python
res["messages"]
```

```python


for message in res["messages"]:
    if isinstance(message, AIMessage) and message.content:
        print(message.content[-1]['text'])
        print()
```
