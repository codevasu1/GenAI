```python
from dotenv import load_dotenv
load_dotenv() 
```

```python
from pydantic import BaseModel
from typing import Annotated
from langchain_groq import ChatGroq
from langgraph.graph import  StateGraph, START, END
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import InMemorySaver
```

```python
class ChatState(BaseModel):
    messages:Annotated[list, add_messages]
```

```python
llmGroq = ChatGroq(model="openai/gpt-oss-120b")
```

```python
def chatBotNode(state:ChatState) -> ChatState:
    res = llmGroq.invoke(state.messages)
    state.messages = [res]
    return state
```

```python
memory = InMemorySaver()
```

```python
graph = StateGraph(ChatState)
graph.add_node("chatbot", chatBotNode)


graph.add_edge(START, "chatbot")
graph.add_edge("chatbot", END)

graph = graph.compile(checkpointer=memory)
```

```python
from IPython.display import Image
Image(graph.get_graph().draw_mermaid_png())
```

```python
res = graph.invoke(
    {"messages": [{"role":"user","content":"Who is their president?"}]},
    {"configurable":{"thread_id":"5r34tejg5gy4jg4099hm5op6bh5rtg54"}}
    )
```
