#### Manage History

```python
history = []

while True:
    query = input("User: ")  
    print("User: ", query)

    if query.lower() in ["bye", "exit", "", "quit"]:
        print("Bye 👋")
        break
    
    history.append({"role":"user", "content": query})

    res = llm.invoke(history)
    # Safely extract text from LangChain content
    if isinstance(res.content, list):
        # Extract text blocks (skips extra metadata like 'extras' or signatures)
        text_content = "".join(
            block["text"] for block in res.content if isinstance(block, dict) and "text" in block
        )
    else:
        text_content = res.content

    print("Agent:", text_content, "\n")

    #Append assistant response with "ai" role
    history.append({"role": "ai", "content": text_content})


```
